# Part 1 → Theory Revision (Quick Recap)    

### What is poisson Distribution?   

Poisson Distribution is a discrete probability distribution that predicts how many times an event will happen in a fixed interval of time or space. 

Here Event exact time cannot be important, only how many times an event occur it's important.   

**Example** 
- Numbere of website visitors in 1 hour.    
- Number of calls received in 10 minutes.   
- Number of network requests per second.    
- Number of manufacturing defects in one batch. 
- Number of patients arriving in one hour.  

### Why is it called an Event Distribution? 

Poisson Distribution is called an Event Distribution because it focuses on counting events. 

It answers questions like:  
- How many customers will arrive today? 
- How many server requests will come in one minute? 
- How many accidents may happen this week?  

It predicts the count of events, not the exact time of each event.  

### Rare Events 

Poisson Distribution is mainly used for rare events.    

A rare event means: 
- It does not happen very often.   
- But it can happen at any time.    

Examples    
- Bank fraud
- Server failure    
- Machine breakdown 
- Car accident  
- Hospital emergency    
- Network attack    

These events are uncommon, but they are very important to predict.  

### Lambda (λ)  

Lambda (λ) is the important parameter in Poisson Distribution.  

It represents the average number of events expected in a fixed interval.    

Example 

suppose a website gets: 

20 visitors per minute  

then    

``` 
λ = 20  
``` 

This means the expected average is 20 events per minute.


### Mean    

the Mean is the average number of events.   

**Formula** 

``` 
Mean = λ    
``` 

If  

``` 
λ = 8   
``` 

Then, Average events = 8    

### Variance    

Variance measures how much the event count changes around the average.  

One special property of Poisson Distribution is:    

``` 
Variance = λ    
``` 

This means: 

``` 
Mean = Variance = λ 
``` 

This is one of the easiest ways to identify Poisson data.   

### Event Rate  

The Event Rate tells us how frequently events happen in a fixed interval.   

Examples    
- 15 customers per hour 
- 200 requests per second   
- 3 accidents per day   
- 5 defects per machine     

The event rate is represented by λ. 

A Large λ means events happen more frequently.  

### Time / Space Interval   

Poisson Distribution always works within a fixed interval.  

The interval can be based on:   

Time    
- per second.   
- per Minute.   
- Per hour. 
- per day.  

Space   
- per kilometer.    
- per page. 
- per machine.  
- per factory.  
- per hospital ward.    

The interval must remain fixed while analyzing the data.    

### When to use Poisson Distribution?   

Use Poisson Distribution when all these conditions are true:    
- You are counting the number of events.    
- Events happen in a fixed time or space interval.  
- Events occur independently.   
- The average event rate (λ) remains approximately constant.
- Events are counted as 0, 1, 2, 3, ...(discrete values).   

If these conditions are not satisfied, another distirbutioon may be more sutiable.  

### Difference Between Binomial and Poisson 

| Binomial Distribution        | Poisson Distribution         |
| ---------------------------- | ---------------------------- |
| Fixed number of trials       | Fixed time or space interval |
| Success or Failure           | Count of events              |
| Uses **n** and **p**         | Uses **λ (Lambda)**          |
| Predicts number of successes | Predicts number of events    |
| Example: Coin Toss           | Example: Website Visitors    |

````python
plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x="Requests",
    bins=15,
    stat="count",
    color="steelblue",
    edgecolor="black"
)

plt.title("Histogram of Poisson Distribution")
plt.xlabel("Number of Requests")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()
````

### 2. PMF PLot 

Poisson is Discrete Distirbution    

that's why there graph is **PMF (Probability Mass Funciton).**      

Histogram only observation views.   

But PMF probability views.

````python
from scipy.stats import poisson

x = np.arange(
    df["Requests"].min(),
    df["Requests"].max()+1
)

lam = df["Requests"].mean()

pmf = poisson.pmf(x, mu=lam)

plt.figure(figsize=(10,5))

plt.stem(
    x,
    pmf,
    basefmt=" "
)

plt.title("Poisson PMF")
plt.xlabel("Number of Requests")
plt.ylabel("Probability")

plt.grid(alpha=0.3)

plt.show()
````

### 3. Histogram + PMF  


````python
plt.figure(figsize=(10,5))

sns.histplot(
    df["Requests"],
    stat="probability",
    bins=15,
    color="skyblue",
    edgecolor="black",
    alpha=0.6
)

plt.plot(
    x,
    pmf,
    "o-r",
    linewidth=2,
    label="Poisson PMF"
)

plt.title("Histogram with Poisson PMF")
plt.xlabel("Requests")
plt.ylabel("Probability")

plt.legend()

plt.grid(alpha=0.3)

plt.show()
````

# Part 6 → Formula Verification

## Objective

This section verifies the Poisson Distribution formulas using both manual calculation and the SciPy library.

It helps us confirm that the generated dataset follows the Poisson Distribution correctly.

---

## 1. Verify Mean

### Theory

The mean of a Poisson Distribution is equal to **λ (Lambda)**.

### Formula

$$
\mu=\lambda
$$

### Practical Check

Calculate the mean from the dataset and compare it with the Lambda value used during data generation.

### Expected Result

If

```python
lam = 8
```

then

```text
Mean ≈ 8
```

A small difference is normal because the dataset is randomly generated.

---

## 2. Verify Variance

### Theory

In a Poisson Distribution,

**Variance is also equal to λ.**

### Formula

$$
\sigma^2=\lambda
$$

### Practical Check

Calculate the variance of the dataset.

Compare it with the Lambda value.

### Expected Result

```text
Variance ≈ Mean ≈ λ
```

If the variance is close to the mean, the dataset behaves like a Poisson Distribution.

---

## 3. PMF Formula

The PMF (Probability Mass Function) calculates the probability of getting exactly **x events**.

### Formula

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}
$$

### Symbol Meaning

| Symbol | Meaning |
|---------|---------|
| $P(X=x)$ | Probability of getting exactly x events |
| $\lambda$ | Average number of events |
| $x$ | Number of observed events |
| $e$ | Euler's Number (≈ 2.71828) |
| $x!$ | Factorial of x |

---

## 4. Verify using SciPy

SciPy provides a built-in function to calculate the Poisson PMF.

```python
poisson.pmf()
```

### Why Use SciPy?

- Fast calculation
- Accurate results
- Industry standard
- Used in Data Science and AI/ML projects

---

## 5. Manual vs SciPy

Calculate the probability using:

- Manual PMF formula
- `scipy.stats.poisson.pmf()`

Compare both results.

### Expected Result

Both values should be almost identical.

If they match, the manual calculation is correct.

---

## 6. Theory vs Practical

Finally, compare the theoretical formulas with the practical dataset.

| Property | Theory | Practical |
|----------|---------|-----------|
| Mean | λ | Calculated Mean |
| Variance | λ | Calculated Variance |
| PMF | Manual Formula | SciPy PMF |

### Final Observation

If

- Mean ≈ λ
- Variance ≈ λ
- Manual PMF ≈ SciPy PMF

then the dataset successfully follows the Poisson Distribution.

The highest point of the graph shows the **most probable number of events**.

Usually, the peak is close to **λ (Lambda)**.

The graph shows how often different event counts occur.

You can estimate Lambda by looking at the graph.

- Small λ → Graph stays on the left.
- Large λ → Graph shifts to the right.

This gives a quick idea about the average event rate.

Most probabilities are concentrated around **λ**.

As we move away from λ, the probability decreases.

This tells us which event counts are most likely.

---

The tail contains events with very low probability.

These events happen rarely but are still possible.

Examples:

- Sudden traffic spike

## 6. Rare Events

Rare events appear in the tail of the graph.

They have a very small probability but are important in real-world applications like fraud detection and anomaly detection.

- Is the graph shifted left or right?
- Is the spread small or large?

# Part 10 → Key Takeaways

## Important Points

- Poisson Distribution works with **event counts**.
- Events occur in a fixed time or space interval.
- Lambda (λ) controls the average number of events.
- Mean and Variance are equal.
- It is a discrete probability distribution.

---

## Common Mistakes

❌ Using Poisson for continuous data.

✔️ Use it only for count data.

---

❌ Ignoring Lambda.

✔️ Lambda controls the entire distribution.

---

❌ Using Poisson when event rate changes.

✔️ Event rate should remain constant.

---

❌ Ignoring independent events.

✔️ Events should be independent.

---

## When to Use

Use Poisson Distribution when:

- Counting events
- Fixed time or space interval
- Independent events
- Constant average event rate

Examples:

- Website requests
- Customer arrivals
- Phone calls
- Manufacturing defects
- Server errors

---

## When Not to Use

Do not use Poisson Distribution when:

- Data is continuous.
- Probability changes over time.
- Events are dependent.
- Number of trials is fixed with only two outcomes (Use Binomial instead).

---

## Quick Revision Notes

- Poisson Distribution models **event counts**.
- λ represents the average number of events.
- Mean = Variance = λ.
- PMF calculates the probability of exactly **x events**.
