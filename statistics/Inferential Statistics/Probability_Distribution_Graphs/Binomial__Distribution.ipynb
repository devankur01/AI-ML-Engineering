# part 1. Theory Revision (Quick Recap)

## What is Binomial Distribution?

A **Binomial Distribution** is a **discrete probability distribution** that calculates the probability of getting a fixed number of **successes** in a fixed number of **independent trials**.

It is used when each trial has only **two possible outcomes**:

- Success
- Failure

---

## Why is it called "Binomial"?

The word **"Binomial"** means **two possible outcomes**.

For every trial, there are only two possible results:

- Success (1)
- Failure (0)

### Examples

- Coin Toss → Head / Tail
- Email → Spam / Not Spam
- Customer → Buy / Not Buy
- Medical Test → Positive / Negative

---

## Bernoulli Trial

A **Bernoulli Trial** is a single experiment that has only **two possible outcomes**.

### Example

One coin toss

- Head → Success
- Tail → Failure

A **Binomial Distribution** is simply a collection of many **Bernoulli Trials**.

---

## Random Variable (X)

The random variable **X** represents the **number of successes**.

### Example

Suppose a coin is tossed **10 times**.

```text
Head = Success
Tail = Failure

Result

H H T H T T H H T H
```

Number of Heads (Successes)

```text
X = 6
```

Here, the random variable **X** is **6** because we got **6 successes**.

---

## Conditions of Binomial Distribution

A Binomial Distribution is valid only when all four conditions are satisfied.

### 1. Fixed Number of Trials

The total number of trials (**n**) must be fixed.

Example:

- Toss a coin 10 times.
- Check 100 products for defects.

---

### 2. Two Possible Outcomes

Each trial must have only two possible outcomes.

- Success
- Failure

Example:

- Pass / Fail
- Yes / No
- Buy / Not Buy

---

### 3. Constant Probability

The probability of success (**p**) must remain the same in every trial.

Example:

If the probability of getting a Head is **0.5**, then it should remain **0.5** for every coin toss.

---

### 4. Independent Trials

Every trial should be independent.

This means the result of one trial should **not affect** the next trial.

Example:

Getting a Head in one coin toss does not change the probability of getting a Head in the next toss.

---

## Parameters

| Symbol | Meaning |
|--------|---------|
| **n** | Number of trials |
| **x** | Number of successes |
| **p** | Probability of success |
| **q = 1 - p** | Probability of failure |

---

## Probability Formula

$$
P(X=x)=\binom{n}{x}p^x(1-p)^{\,n-x}
$$

### Formula Meaning

- **P(X = x)** → Probability of getting exactly **x** successes.
- **n** → Total number of trials.
- **x** → Number of successes.
- **p** → Probability of success.
- **1 - p** → Probability of failure.
- **\(\binom{n}{x}\)** → Number of possible combinations.

---

## Mean

The **Mean** represents the **expected number of successes**.

### Formula

$$
\mu=np
$$

---

## Variance

Variance measures **how much the data is spread around the mean**.

### Formula

$$
\sigma^2=np(1-p)
$$

---

## Standard Deviation

Standard Deviation shows the **actual spread** of the data.

### Formula

$$
\sigma=\sqrt{np(1-p)}
$$

---

## Shape of Binomial Distribution

The shape depends on the value of **p**.

### When p = 0.5

- Distribution is approximately **Symmetric**.

### When p < 0.5

- Distribution is **Right Skewed**.
- Most values are on the left side.

### When p > 0.5

- Distribution is **Left Skewed**.
- Most values are on the right side.

As the number of trials (**n**) becomes larger, the Binomial Distribution becomes closer to a **Normal Distribution**.

---

3. Verify Standard Deviation

$$
\sigma = \sqrt{np(1-p)}
$$

Standard deviation actual spread same unit measured.

````python
theoretical_std = np.sqrt(n * p * (1 - p))

practical_std = df["Successes"].std()

print(f"Theoretical SD : {theoretical_std:.2f}")
print(f"Practical SD   : {practical_std:.2f}")
````

**Interpretation**

#### 4. PMF Formula Verification    

Theory Formula  

$$
P(X=x)=\binom{n}{x}p^x(1-p)^{\,n-x}
$$

Manual Calculation  

Suppose     

``` 
n = 10

p = 0.5

x = 5   
``` 


````python
from math import comb

x = 5

manual_probability = (
    comb(n, x)
    * (p ** x)
    * ((1 - p) ** (n - x))
)

print(f"Manual PMF : {manual_probability:.5f}")
````

````python
# Verify with scipy.stats.binom.pmf 

from scipy.stats import binom

scipy_probability = binom.pmf(x, n, p)

print(f"SciPy PMF : {scipy_probability:.5f}")
````

````python
# Compare

print(f"Difference : {abs(manual_probability - scipy_probability):.10f}")
````

**Interpretation**  

If difference   

``` 
0.0000000000    
``` 
Then very close,    

so, Manual formula are corrected

- Small differences between theory and practical values are normal because the data is randomly generated.
- Formula verification helps ensure that the dataset follows the expected Binomial Distribution.

## Part 7 → Graph Interpretation    

**AI/ML Checklist While Reading a Binomial Distribution Graph**

 #### ✅ 1. Is the graph symmetric or skewed?

What to check?

Look at the overall shape of the graph.

- Is the left side similar to the right side?
- Or is one side longer than the other?

**Interpretation**

- Symmetric → Data is balanced. Usually when p ≈ 0.5.
- Right Skewed → Most values are small, tail is on the right. Usually when p < 0.5.
- Left Skewed → Most values are large, tail is on the left. Usually when p > 0.5.

**AI/ML Importance**

The shape tells you whether the data is balanced or biased. This helps in choosing the right model and preprocessing technique.

#### ✅ 2. Where is the peak?

What to check?

Find the highest point (or tallest bar) in the graph.

**Interpretation**

The peak shows the most common outcome.

This is the result that is expected to occur most often.

**AI/ML Importance**

The peak helps identify the most likely prediction or the most common behavior in the data.

#### ✅ 3. What is the most probable outcome?

What to check?

Identify which value has the highest probability.

**Interpretation**

This value is the one that is most likely to happen.

Example:

If the highest probability is at 5 successes, then getting 5 successes is the most expected outcome.

**AI/ML Importance**

This helps estimate the expected result in real-world prediction problems.

✅ 4. Where is the mean?

What to check?

Locate the average value of the distribution.

**Interpretation**

- In a symmetric graph, the mean is usually near the center.
- If the graph is skewed, the mean shifts toward the longer tail.

**AI/ML Importance**

The mean represents the expected average prediction and is useful for understanding the center of the data.

#### ✅ 5. Is the spread small or large?

What to check?

Observe how wide the graph is.

**Interpretation**

- Small Spread → Data is close to the mean and more consistent.
- Large Spread → Data is more scattered and less consistent.

**AI/ML Importance**

- Small spread → More stable predictions.
- Large spread → Higher uncertainty and more difficult predictions.

#### ✅ 6. Which outcomes have the highest probability?

What to check?

Look for the region where the graph is tallest.

**Interpretation**

These outcomes are the most likely to occur.

The model will usually predict values from this region.

**AI/ML Importance**

This helps identify the most common user behavior, most common events, or most likely predictions.

#### ✅ 7. Are extreme outcomes rare?

What to check?

Look at the bars near the beginning and end of the graph.

**Interpretation**

If those bars are very small,

it means extreme outcomes have very low probability.

**AI/ML Importance**

Rare events often represent:

Fraud Detection
Equipment Failure
Medical Emergencies
Network Attacks
Anomaly Detection

Even though they are rare, they are often the most important events in AI/ML.

#### ✅ 8. Does the graph match the theoretical Binomial Distribution?

What to check?

Compare the practical graph with the expected Binomial shape.

Check:

- Shape
- Peak
- Spread
- Mean

**Interpretation**

If everything is close to the theoretical distribution,

your generated dataset is statistically correct.

**AI/ML Importance**

Before training a model, Data Scientists verify that the dataset behaves as expected. If the graph does not match the theory, there may be a problem with the data generation process or assumptions.

# When Should You Use Binomial Distribution?

Use Binomial Distribution when:

- The number of trials is fixed.
- Each trial has only two possible outcomes.
- The probability remains constant.
- Every trial is independent.

If these conditions are not satisfied, use another probability distribution such as **Normal Distribution** or **Poisson Distribution**.

---

# Common Mistakes

❌ Using Binomial Distribution for continuous data.

✔️ Binomial Distribution works only with **discrete binary outcomes**.

---

❌ Using Binomial Distribution when the probability changes in every trial.

✔️ The probability should remain **constant**.

---

❌ Ignoring the independence of trials.

✔️ Every trial should be **independent**.
