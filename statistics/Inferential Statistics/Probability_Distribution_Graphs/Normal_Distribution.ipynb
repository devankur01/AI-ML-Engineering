#### What is Normal Distribution?   

Normal Distribution is the most common probability distribution in Statistics and AI/ML. 
It describes how continuous numercial data is distributed around its **mean(average)**. 
Most values are close to the center, and fewer values appear as we move away from the center.   

Example:    
- Employee Salary.  
- Sutdent Marks.    
- Human Height. 
- Weight.   
- IQ Score. 
- Measurement Errors.   

Many real-world datasets are approximately normally distributed.    

#### Why is it called Gaussian Distribution?    

Normal Distribution is also called the Gaussian Distribution because it was studied and popuarized by the mathematician **Carl Friedirich Gauss.**  

Whenever data naturally clustes around the center and decreases smoothly on both sides, it follows a Gaussian Distribution. 

Both names refer to the **same distribution.**

#### Mean(μ)    

the **Mean** is the center of the distribution. 

Formula 

$$
\mu=\frac{\sum x}{n}
$$  

Graph Interpretation    
- Mean is located at the highest point of the curve.    
- In aa perfect Normal Distribution, the **Mean = Median = Mode.**  
- If the mean shifts, the entire curve moves left or right. 

#### Standard Deviation(σ)  
standard Deviation measures how much the data spread around the mean.   

Formula 

$$
\sigma=\sqrt{\frac{\sum(x-\mu)^2}{N}}
$$  

Graph Interpretation    
- Small σ → Narrow and tall curve.    
- large σ → Wide and short curve. 
- Higher σ means data is more spread out. 

#### Probability Density Function(PDF)  
The PDF describes how the dataa is distributed. 

Formula 

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$  

Important Points    
- PDF does not give the probability of one exact value. 
- It shows which **range of values** is more likely.    
- The **higher the curve**, the more data is concentrated in that region.   
- The **total area under the curve is always 1 (100%).**    

#### Bell Curve 
The graph of a Normal Distribution is called a **Bell Curve** because it looks like a bell. 

Properties  
- Symmetric 
- One Peak  
- Mean at the center    
- Equal left and right sides
- Smooth curve  
- No shape edges    

Graph Interpretation    
- Highest point = Highest data concentration.   
- Ends of the curve = Rare values.  

#### 68–95–99.7 Rule    
This rule explains where most of the data lies. 

| Range  | Percentage of Data |
| ------ | -----------------: |
| μ ± 1σ |      About **68%** |
| μ ± 2σ |      About **95%** |
| μ ± 3σ |    About **99.7%** |

Meaning 
- Around **68%** of values are within one standard Deviation.
- Around **95%** of values are within two standard deviations.  
- Around **99.7%** of values are within three Standard Deviations.  
- Values outside **±3σ** are very rare and may be considered potential outliers.    

#### How to Read a Normal Distribution Graph    
Whenever you see a Normal Distribution graph, always check these points in order.   

1. Center   

Where is the Peak?  

→ This is approximately the Mean (μ).   

2. Shape    

Is the graph symmetric? 

→ If yes, it is close to Normal Distribution.

3. Spread   

Is the curve narrow or wide?    

- Narrow → Small Standard Deviation.    
- Wide → Large Standard Deviation.  

4. Concentration    

Where is the curve highest? 

→ That is where most of the data is concentrated.   

5. Tails    

Look at both ends of the curve. 

Very few observations should appear there.  

6. Area Under Curve 

Remember:   

The total are under the curve is always 1 (100%).   

The area Between two points represents the probability of the data falling within that range.   

#### AI/ML Connection   

Normal Distribution is important because many machine Learning techniques assume that data is approximately normal. 

It is commonly used in: 
- Exploratory Data Analysis(EDA).

### 4. Mean, Median & Standard Deviation

````python
mean=df["Salary"].mean()
median=df["Salary"].median()
std=df["Salary"].std()

plt.figure(figsize=(12,6))

sns.histplot(
    data=df,
    x="Salary",
    bins=30,
    stat="density",
    color="lightblue",
    alpha=0.5
)

sns.kdeplot(
    data=df,
    x="Salary",
    color="blue",
    linewidth=2
)

plt.axvline(mean,color="red",label="Mean",linewidth=2)

plt.axvline(median,color="green",label="Median",linewidth=2)

plt.axvline(mean+std,color="orange",linestyle="--",label="+1 SD")

plt.axvline(mean-std,color="orange",linestyle="--",label="-1 SD")

plt.legend()

plt.grid(alpha=0.3)

plt.show()
````

**Interpretation**

Mean ≈ Median

↓

Data almost symmetric

Large distance

↓

Skewed data

±1 SD

↓

Around 68% data

### 5. PDF Curve    


````python
from scipy.stats import norm

mean=df["Salary"].mean()
std=df["Salary"].std()

x=np.linspace(
    df["Salary"].min(),
    df["Salary"].max(),
    1000
)

pdf=norm.pdf(x,mean,std)

plt.figure(figsize=(11,5))

sns.histplot(
    data=df,
    x="Salary",
    stat="density",
    bins=30,
    alpha=0.5,
    color="skyblue"
)

plt.plot(
    x,
    pdf,
    color="red",
    linewidth=3,
    label="Normal PDF"
)

plt.legend()

plt.grid(alpha=0.3)

plt.show()
````

**Interpretation**

Histogram

↓

Actual Data

Red Curve

↓

Theoretical Normal Distribution

Almost overlap

↓

Data follows Normal Distribution

### 6. Q-Q Plot

````python
from scipy import stats

plt.figure(figsize=(6,6))

stats.probplot(
    df["Salary"],
    dist="norm",
    plot=plt
)

plt.title("Q-Q Plot")

plt.grid(alpha=0.3)

plt.show()
````

**Interpretation**

### 7. Shapiro-Wilk Test    

````python
from scipy.stats import shapiro

stat,p=shapiro(df["Salary"])

print("Statistic :",stat)

print("P-value :",p)
````

**Decision**    

``` 
p > 0.05

Normal  
``` 

``` 
p < 0.05

Not Normal  
``` 

here our data is normal because our p value is greater then 0.05 is P-value : 0.6272578392965296.

### 8. Anderson-Darling Test

````python
from scipy.stats import anderson

result=anderson(df["Salary"])

print(result.statistic)

print(result.critical_values)

print(result.significance_level)
````

**Decision**

Statistic

<

Critical Value

↓

Normal

### 9. Kolmogorov-Smirnov Test

````python
from scipy.stats import kstest
from scipy.stats import norm

mean = df["Salary"].mean()
std = df["Salary"].std()

stat, p = kstest(
    df["Salary"],
    norm.cdf,
    args=(mean, std)
)

print("Statistic :", stat)
print("P-value :", p)
````

**Decision**    

p > 0.05

Normal
