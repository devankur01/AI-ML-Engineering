# All About Regression Analysis 

### 1. What is Regression?  

Regression is a statistical and ML method used to understand and predict a numerical value. 

Simple words:       

Regression tells us how one or more variables affect a numerical target and helps us predict that target.   

**Real-world example**  

Suppose:    
- ```Experience``` → 2, 4, 6, 8 years.  
- ```Salary``` → ₹35K, ₹50K, ₹70K, ₹90K.    

We can study: 

Does salary increase when experience increases?   

And then Predict:   

If experience = 7 years, what salary can we expect? 

That's **Regression.**  

### 2. Independent Variable & Dependent Variable    

In regression mainly two types variables.   

**Independent Variable (X)**    

these are the input/features those explain or predicts the target.  

Examples:   
- Experience.   
- House Area.   
- Advertising Spend.    
- Number of Rooms.  

**Dependent Variable (Y)**  

these are those whose we can predicts.  

Examples:   
- Salary.   
- House Price.  
- Sales.    
- Revenue.  

**Easy Rule**   

**X = Input → Y = Output**  

Example:    

``` 
Experience (X) ───────► Salary (Y)
       Input             Target 
``` 

### 3. Regression vs Correlation    

These boths are related, but don't same.    

**Correlation** 

Correlation tells:  

**How strongly two variables are related.**

Example:    

``` 
Experience ↔ Salary 
``` 

It mainly measures the **strength and direction of relationship.**  

Correlation coefficient:    

−1≤r≤1  

- ```r = +1``` → Strong positive relationship.  
- ```r ≈ 0``` → Little/no linear relationship.  
- ```r ≈ -1``` → Strong negative relationship.  

**Regression**  

Regression goes one step further:

**It uses the relationship to estimate/predict Y from X.**  

``` 
Correlation → Relationship

Regression → Relationship + Prediction  
``` 

**Important**   

**Correlation does not prove causation.**   

A strong correlation does not automatically mean X causes Y.

### 4. Linear Relationship

X and Y relationship represents one mathematical line.  

Example:    

```

### 5. Simple Linear Regression 

If we have only 1 Independent variable we can says that Simple Linear Regression.   

Example:    

``` 
Experience → Salary 
``` 

Formula:    

Y=β0​+β1​X+ϵ    

Where:  

- Y = predicted/target value
- X = input feature
- β₀ = intercept
- β₁ = slope/coefficient
- ε = error/noise

**Easy Understanding**  

``` 
Prediction = Intercept + Slope × Input  
``` 

### 6. Multiple Linear Regression 

When multiple independent variables predicts the target, we use Multiple Linear Regression. 

Example:  

``` 
Area
Rooms
Location
Age
   ↓
House Price 
``` 


Formula:  

Y=β0​+β1​X1​+β2​X2​+⋯+βn​Xn​+ϵ  

Example:  

Price=β0​+β1​(Area)+β2​(Rooms)+β3​(Age) 

Simple Difference 

| Type                       | Features    |
| -------------------------- | ----------- |
| Simple Linear Regression   | 1 feature   |
| Multiple Linear Regression | 2+ features |

### 7. Regression Equation  

Regression equation represents model mathematical relationship. 

For simple regression:  

Y^=b0​+b1​X 

Where:  

- Ŷ = predicted Y
- b₀ = intercept
- b₁ = slope
- X = input 

Example:  

Salary=30000+8000(Experience) 

If: 

``` 
Experience = 5  
``` 

then: 

``` 
Salary = 30000 + 8000 × 5
       = 70000  
``` 

so predicted salary = ₹70,000.  


### 8. Slope  

Slope tells us how much Y changes when X increases by 1 unit. 

Formula:  

$$
b_1 = \frac{\Delta Y}{\Delta X}
$$  

**Positive Slope**  

``` 
X ↑ → Y ↑ 
``` 

Example : 

Experience increases → Salary tends to increases. 

**Negative Slope**  

``` 
X ↑ → Y ↓ 
``` 

Example:  

Price increases → Demand may decrease.  

**Zero/near-zero Slope**  

```
X changes → Y changes very little 
``` 

**Interpretation**  

If: 

b1​=5000  

then: 

For every 1-unit increases in X, predicted Y increases by about 5000 units, assuming other variable stay constant.  

### 9. Intercept  

Intercept is the predicted value of Y when X = 0. 

Formula:  

b0​=value of Y when X=0 

Example:  

Salary=30000+8000(Experience)

When: 

``` 
Experience = 0  
``` 

then: 

``` 
Salary = ₹30,000  
``` 

So, ```₹30,000``` is the intercept.

**Important** 

Intercept mathematically important, but real-world meaning only useful when **X = 0 realistic.**  

### 10. Residuals 

Regression prediction usually not exact.  

Actual value and predicted value difference is Residual.  

Residual=Actual−Predicted 

Example:  

``` 
Actual Salary     = ₹75,000
Predicted Salary  = ₹70,000 
``` 

Then: 

Residual=75000−70000=5000 

**Interpretation**  

- Positive residual → model predicted too low
- Negative residual → model predicted too high
- Residual ≈ 0 → prediction close to actual 

**Visual idea** 

``` 
Actual Point ●
             │
             │ Residual
             │
Regression ──●────────
          Predicted 
``` 

Residual later model diagnostics important. 

### 11. Regression Line 

Regression line whose line that represents overall data patterns. 

Normally models these line find because of predictions errors overall minimum.

For Ordinary Least Squares(OLS), model commonly minimizes:  

∑(Y−Y^)2

Meaning:  

Actual and predicted values squared difference as small as possible make it.  

that's why regression line often **best-line** says.  

### 12. Regression Assumptions  

Before Regression model we can make we firstly checks assumption its useful.  

**1.Linearity** 

X and Y realationship approximately lineared. 

``` 
Good:
●
  ●
    ●
      ●

Bad:
●
  ●
     ●
        ●
      ● 
``` 

**2. Independence** 

Observations/residuals ideally independent. 

One observation error another observation error cannot strongly influence.

**3. Homoscedasticity** 

Residuals spread roughly constant.  

Simple meaning: 

**prediction errors X different levels suddenly not more wide/narrow.** 

**4. Normality of Residuals** 

For classic statistics inference, residuals are often expected to be approimately normally distributed. 

Important:  

**Target variable not mandatory normal variable.**  

Residuals distribution martters, especially when we want reliable confidence intervals and hypothesis tests.  

**5. No Server Mutlicollinearity**  

Multiple Linear Regression main features one or other extremely not correlated to each other. 

Example:  

```

     ↕

``` 

If both almost same information gives, then coeffiecents unstable.

**6. Outliers & Influential Points**  

Extreme observations regression line are strongly affects.  

Example:  

```

          ●
                 ●

```

Tools:  
- Train/Test Split.

- R².

Regression is not only predictions technique they are also statisticall or ML boths.

```

**Outlier Mathematically varify**   

Boxplot ka basic rule:

``` 
IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR    
``` 


````python
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
````

````python
# Now outlier

price_outliers = df[
    (df["Price"] < lower_bound) |
    (df["Price"] > upper_bound)
]

print("Number of Price Outliers:", len(price_outliers))
````

````python
# With All features checks

def find_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\n{column}")
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outliers:", len(outliers))


for column in ["Area", "Age", "Location_Score", "Price"]:
    find_outliers(column)
````

````python
# Build simple linear Regression 

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Feature and Target
X = df[["Area"]]
y = df["Price"]

# 2. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# 3. Create Model
model = LinearRegression()

# 4. Train Model
model.fit(X_train, y_train)

# 5. Model Parameters
slope = model.coef_[0]
intercept = model.intercept_

print("Slope:", slope)
print("Intercept:", intercept)

# 6. Regression Equation
print(
    f"Regression Equation: Price = "
    f"{slope:.2f} × Area + {intercept:.2f}"
)

# 7. Prediction
y_pred = model.predict(X_test)

# 8. Actual vs Predicted
comparison = pd.DataFrame({
    "Actual_Price": y_test.values,
    "Predicted_Price": y_pred
})

display(comparison.head(10))

# 9. Visualization
plt.figure(figsize=(8, 5))

plt.scatter(
    X_test["Area"],
    y_test,
    alpha=0.6,
    label="Actual"
)

plt.plot(
    X_test["Area"],
    y_pred,
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Simple Linear Regression: Area vs Price")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
````

````python
# Part 6 = Model Evaluation

import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Predictions
y_pred = model.predict(X_test)

# Errors
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

# R²
r2 = r2_score(y_test, y_pred)

# Adjusted R²
n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1 - (
    (1 - r2) * (n - 1)
    / (n - p - 1)
)

print("Model Evaluation")
print("-------------------------")
print("MAE          :", mae)
print("MSE          :", mse)
print("RMSE         :", rmse)
print("R² Score     :", r2)
print("Adjusted R²  :", adjusted_r2)
````

MSE errors we can squared.  

```MSE = average((Actual - Predicted)²)```

RMSE - how the average error, with extra penalty for large errors?

they says:  

Model target variable variation how the part explained.

Simple Interpretation:

R² range    

Generally:  

``` 
1.0  → Perfect fit
0.8  → Strong explanatory power
0.5  → Moderate
0.0  → No improvement over mean baseline
< 0  → Very poor model  
``` 

⚠️ R² = 0.82 ka means model cannot 82% accurate.

**Adjusted R²** 

Adjusted R² mainly **Multiple Linear Regression** in useful.    

Normal R²:  

More features we can adds generally not decreases.  

Problem:    

```
Feature add
   ↓
R² increase
   ↓
But feature actually not we useful  
```

Adjusted R² unnecessary features penalty adds.  

Formula:    

``` 
Adjusted R² =
1 - [(1 - R²)(n - 1) / (n - p - 1)] 
``` 

Where:  

``` 
n = number of observations
p = number of features  
``` 

Code:   

```py   
n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1 - (
    (1 - r2) * (n - 1)
    / (n - p - 1)
)

print("Adjusted R²:", adjusted_r2)  
``` 

Important   

Simple Linear Regression ```p=1```, so R² and Adjusted R² usually closed.   

**Multiple Linear Regression in Adjusted R² much more useful.**

````python
# Build Multiple Regression Part --> 8

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)   

# Define Features and Target    

X = df[
    [
        "Area",
        "Bedrooms",
        "Age",
        "Location_Score"
    ]
]

y = df["Price"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())

# Train-Test Split Meaning 80% --> Training and 20% --> Testing

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)  

# Create Multiple Linear Regression Model   

model_multi = LinearRegression()

print("Multiple Linear Regression model created.")  

# Train the Model   

model_multi.fit(X_train, y_train)

print("Model training completed.")  

# Get Intercept(β₀)

print("Intercept:", model_multi.intercept_) 


# Get Feature Coefficients

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model_multi.coef_
})

coefficients

# Display Regression Equation 
'''
Conceptually :

Price = β₀
      + β₁(Area)
      + β₂(Bedrooms)
      + β₃(Age)
      + β₄(Location_Score)
'''  

print("Regression Equation:\n")

print(
    f"Price = {model_multi.intercept_:.2f}"
    f" + ({model_multi.coef_[0]:.2f} × Area)"
    f" + ({model_multi.coef_[1]:.2f} × Bedrooms)"
    f" + ({model_multi.coef_[2]:.2f} × Age)"
    f" + ({model_multi.coef_[3]:.2f} × Location_Score)"
)


# Make Predictions  

y_pred_multi = model_multi.predict(X_test)

print("First 10 Predictions:")
print(y_pred_multi[:10])    

# Actual vs Predicted Values

comparison_multi = pd.DataFrame({
    "Actual_Price": y_test.values,
    "Predicted_Price": y_pred_multi
})

comparison_multi.head(10)   

# Calculate Prediction Error

comparison_multi["Error"] = (
    comparison_multi["Actual_Price"]
    - comparison_multi["Predicted_Price"]
)

comparison_multi.head(10)



````
