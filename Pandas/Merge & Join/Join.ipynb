Join is used to combine DataFrames based on their index.    

Merge → On the basis of Column.     

Join → On the basis of Index.   

Syntax : df1.join(df2). 

→ df2 add to the df1 on the basis of the index

````python
import pandas as pd
````

````python
# Here we cannot gives the index manully to the Df, It will automatically assign the index to the Df.
employee = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"]
})  

salary = pd.DataFrame({
    "salary": [50000, 60000, 55000, 70000]
})  

result = employee.join(salary) # here the employee Df adds to the salary Df priority is given to the employee Df. 

print(result)
````

````python
# Custom Index. 
employee = pd.DataFrame(
{
    "name": ["Alice", "Bob", "Charlie", "David"]
},
index=[101, 102, 103, 104]
)   

salary = pd.DataFrame(
    {
        "salary": [50000, 60000, 55000, 70000]
    },
    index=[101, 102, 103, 104]
)   

result = employee.join(salary)  
print(result)
````

Types of Join.

- Left join(Default)    

```df1.join(df2)```

Actually:   

```df1.join(df2, how="left")```

same

````python
# Left Join.
employees = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])   

salary = pd.DataFrame({
    "salary": [50000, 60000]
}, index=[101, 102])    


result = employees.join(salary, how="left")  # left join
print(result)      # completely left dataframe table is prefered.
````

````python
# Inner Join.
employees = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])   

salary = pd.DataFrame({
    "salary": [50000, 60000]
}, index=[101, 102])    


result = employees.join(salary, how="inner")  # inner join
print(result)      # only rows with matching indices are included not NaN output show whose completely same indexes only shows in the output.
````

````python
# Right Join.
employees = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])   

salary = pd.DataFrame({
    "salary": [50000, 60000, 349863]
}, index=[101, 102, 104])    


result = employees.join(salary, how="right")  # right join
print(result)      # completely right dataframe table is prefered.
````

````python
# outer Join.
employees = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])   

salary = pd.DataFrame({
    "salary": [50000, 60000]
}, index=[101, 102])    


result = employees.join(salary, how="outer")  # outer join
print(result)      # all rows from both dataframes are included.
````

````python
# Join on Columns.

Employees = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "id": [1,2,3]
})   

Salary = pd.DataFrame({
    "salary": [50000, 60000, 55000],
    "id": [1,2,3]
})  

result = Employees.join(
    Salary.set_index("id"),
    on="id"
)   

print(result)
````

````python
# Multiple DataFrames Join. 

employee = pd.DataFrame({
    "name": ["Alpha", "Rahul", "Aman"]
})  

salary = pd.DataFrame({
    "salary": [50000, 60000, 70000]
})

department = pd.DataFrame({
    "department": ["IT", "HR", "Finance"]
})  

bonus = pd.DataFrame({
    "bonus": [5000, 3000, 7000]
})

result = employee.join(
    [salary, department, bonus]
)   

print(result)
````

````python
# Overlapping Columns Error.

Employee = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
})  

Manager = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alpha", "Theeta", "Gamma"]   
})  

result = Employee.join(Manager, lsuffix="_emp", rsuffix="_mgr")  # here the name in both DataFrame is common that's why we can lsuffix and rsuffix in passes the name is added in the name without this ValueError.

print(result)
````
