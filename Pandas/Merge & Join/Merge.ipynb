Merge means two or more then tables add together on the basis of common column in all the tables.

```python
import pandas as pd

students = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
})  

marks = pd.DataFrame({
    "id": [1,2,3],
    "marks": [85, 90, 95]
})  

result = pd.merge(
    students,
    marks,  
    on="id"                   # here is one of the ways to merge two dataframes based on a common column (id in this case)
)   

print(result)
```

Without common column two or more tables cannot adds togther by the use of merge.

```python
# merges types. 
# Inner Join(Default) --> here's the common columns and rows in the both dataframes only returnes in the resulted output.   
import pandas as pd 
students = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
}) 

marks = pd.DataFrame({
    "id": [2,3,4],  
    "marks": [90, 50, 78],  
})  

result2 = pd.merge(
    students,
    marks,  
    on="id",    
    how="inner",    
    indicator=True
)   

print(result2)
```

```python
# merge type    
# left join --> here's the two tables one is right and second one is left so they join on the bases of common column but in output's shows the whole left table contant.    
import pandas as pd     

students = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
}) 

marks = pd.DataFrame({
    "id": [2,3,4],  
    "marks": [90, 50, 78],  
})  

result3 = pd.merge(
    students,   
    marks,  
    on="id", 
    how="left", 
    indicator=True
)       

print(result3)
```

```python
# Right Join --> here the right table contant whole show's in the output when we add togther two tables.    
import pandas as pd 

students = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
}) 

marks = pd.DataFrame({
    "id": [2,3,4],  
    "marks": [90, 50, 78],  
})      

result4 = pd.merge(
    students,   
    marks,  
    on="id",    
    how="right",    
    indicator=True
)   

print(result4)
```

```python
# Here we can merges two tables on the bases of the common column id and how to merge outer means boths tables data shows in the output table.  

import pandas as pd 

students = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
}) 

marks = pd.DataFrame({
    "id": [2,3,4],  
    "marks": [90, 50, 78],  
})  

result5 = pd.merge(
    students,
    marks,  
    on="id",    
    how="outer",    
    indicator=True   # this is the one of the parameter which shows the data from which table in the output.
)       

print(result5)

```

Inner --> common only.  
Left --> Left Everything.   
Right --> Right Everything.  
Outer --> Everything From Both.

```python
# Differnet column Names.   
import pandas as pd 

students = pd.DataFrame({
    "student_id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"]
}) 

marks = pd.DataFrame({
    "id": [2,3,4],  
    "marks": [90, 50, 78],  
})  

result6 = pd.merge(
    students,   
    marks,  
    left_on="student_id",   
    right_on="id"
)   

print(result6)
```

```python
# Merge two tables on the bases of one more then common column.
import pandas as pd 

employee1 = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alice", "Bob", "Charlie"],    
    "city": ["new-york", "Londan", "Toronto"]
})  

employee2 = pd.DataFrame({
    "id": [1,2,3],  
    "name": ["Alpha", "Theeta", "Gamma"],    
    "city": ["New-zeland", "UK", "India"]
})  

result7 = pd.merge(
    employee1,  
    employee2,  
    on=["id", "city"]
)   

print(result7)   # here's the output is empty due to id and city both are common column in both DataFrames(tables) but city is still mis-match in the both tables, so the output is empty.
```

```python
# here we can merges two tables on the bases of multiple common columns(id and city) and here both id and city are same by thier values in both tables.
import pandas as pd

employee1 = pd.DataFrame({
    "id": [1,2,3],
    "city": ["Delhi","Mumbai","Pune"]
})

employee2 = pd.DataFrame({
    "id": [1,2,3],
    "city": ["Delhi","Mumbai","India"]
})  

result8 = pd.merge(
    employee1,
    employee2,  
    on=["id", "city"]
)   

print(result8)
```

```python
# Duplicates in the common column.  
import pandas as pd 

employee = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ankur", "Rahul", "Aman"]
})  

salary = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["A", "B", "C"],
    "salary": [50000, 60000, 70000]
})  

result9 = pd.merge(
    employee,       # left table
    salary,         # right table
    on="id"
)   

print(result9)   # here the both tables have common name column(name) pandas gets confuse that's why they can left name column as name_x and right name column as name_y.
```

```python
import pandas as pd 

employees = pd.DataFrame({
    "emp_id": [101, 102],
     "department": ["HR", "IT"]
})  

projects = pd.DataFrame({
    "emp_id": [101, 102],
    "department": ["software", "Recruitment"]
})      

# Here the without suffix parameter the common columns whose doesn't merge the two tables another column is department so their left table department columns and right table have the department_x and department_y names that is confusing so we use custom suffixes for that type of name.
result10 = pd.merge(  
    employees,   
    projects,  
    on="emp_id",    
    suffixes=("_emp", "_proj")   # here we can change the default suffixes(_x and _y) to our own suffixes.
)   

print(result10) 

# duplicates row remove by the help of "df.drop_duplicates()" method.
```

```python
# Merge three tables together.
from typing_extensions import final

import pandas as pd 

employees = pd.DataFrame({
    "id": [101, 102, 103],
    "name": ["Alpha", "Rahul", "Aman"]
})  

salary = pd.DataFrame({
    "id": [101, 102, 103],
    "salary": [50000, 60000, 70000]
})  

department = pd.DataFrame({
    "id": [101, 102, 103],
    "department": ["IT", "HR", "Finance"]
})  

# here we want to gets one table like Employee name, salary and department sow we can do merge here.
# Emplyee + Salary in one table.    
temp = pd.merge(
    employees,
    salary,
    on="id"
)   

# Final Employee + Salary + Department in one table.    
final = pd.merge(
    temp,
    department,
    on="id"
)   

print(final)
```

```python
# Validate the merge keys.  
# Normal Merge. 
import pandas as pd

employees = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ankur", "Rahul", "Aman"]
})

salary = pd.DataFrame({
    "id": [1, 2, 2, 3],
    "salary": [50000, 60000, 65000, 70000]
})

r = pd.merge(
    employees,
    salary,
    on="id"
)

print(r)   # here the id 2 is duplicate in the salary table so the output is also duplicate for id 2.
```

```python
# Validate the merge keys.  
# validate types = one-to-one.
# ohters one-to-many, many-to-one, many-to-many. 
import pandas as pd

employees = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ankur", "Rahul", "Aman"]
})

salary = pd.DataFrame({
    "id": [1, 2, 2, 3],
    "salary": [50000, 60000, 65000, 70000]
})

r = pd.merge(
    employees,
    salary,
    on="id",    
    validate="one_to_one"   # here we can validate the merge keys and here the id 2 is duplicate in the salary table so it shows the error.
)

print(r)
```

Concat = here they cannot add the tables on the common column but they can add tables by side by side or upper and lower side one single table. 
Syntax: pd.concat([df1, df2]) --> list inside we can pass.

```python
import pandas as pd 

jan = pd.DataFrame({
    "id": [1, 2],    
    "name": ["Alice", "Bob"]    
})  

feb = pd.DataFrame({
    "id": [3, 4],
    "name": ["ganater", "Charlie"]
})  

result = pd.concat([jan, feb])  

print(result)   
```

```python
# ignore_index = True. 
import pandas as pd

jan = pd.DataFrame({
    "id": [1, 2],    
    "name": ["Alice", "Bob"]    
})  

feb = pd.DataFrame({
    "id": [3, 4],
    "name": ["ganater", "Charlie"]
}) 

result = pd.concat([jan, feb], ignore_index=True, axis=0)  

print(result)   
```

axis = 0 (default) --> Append one DataFrame below another.  
axis = 1 --> Combine columns side by side(Join DataFrames horizontally).    
Exmaple of axis 1 : DataFrame 1 | DataFrame 2.

```python
import pandas as pd 

student = pd.DataFrame({
    "name": ["Ankur", "Rahul"]
})

marks = pd.DataFrame({
    "marks": [90, 85]
})

result = pd.concat(
    [student, marks],
    axis=1
)   

print(result)
```
