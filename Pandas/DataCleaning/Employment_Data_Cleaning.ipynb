# Step 3: Identify and Handle Missing Values

Missing values can affect analysis and machine learning models. Therefore they must be identified and handled.

```python
df.isnull().sum()  # Calculate the number of missing values in each column of the DataFrame
```

```python
(df.isnull().sum() / len(df)) * 100  # Calculate the percentage of missing values in each column of the DataFrame
```

```python
df["Years of Experience"] = df["Years of Experience"].fillna(
    df["Years of Experience"].median()
)  # Fill missing values in the "Years of Experience" column with the median value of that column
```

```python
df["Monthly Salary (INR)"] = df["Monthly Salary (INR)"].fillna(
    df["Monthly Salary (INR)"].median()
)  # Fill missing values in the "Monthly Salary (INR)" column with the median value of that column
```

```python
categorical_columns = [
    "Status",
    "Age Group",
    "Education",
    "Industry",
    "Location",
    "AI Risk"
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])  # Fill missing values in categorical columns with the mode (most frequent value) of each column
```

# Step 4: Remove Duplicate Records

```python
df.duplicated().sum()  # Calculate the number of duplicate rows in the DataFrame
```

```python
df = df.drop_duplicates()  # Remove duplicate rows from the DataFrame
```

# Step 5: Correct Inconsistent Formatting and Text Values

```python
text_columns = [
    "Status",
    "Age Group",
    "Education",
    "Industry",
    "Location",
    "AI Risk"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.title()  # Standardize text columns by converting to string, stripping whitespace, and converting to title case
```

# Step 6: Fix Data Type Issues

```python
df["Date Recorded"] = pd.to_datetime(
    df["Date Recorded"]
) # Convert the "Date Recorded" column to datetime format
```

```python
df.dtypes  # Display the data types of each column in the DataFrame after cleaning
```

# Step 7: Detect and Handle Outliers

```python
df.describe()  # Generate descriptive statistics for the DataFrame, including count, mean, std, min, and max values for numeric columns
```

```python
Q1 = df["Monthly Salary (INR)"].quantile(0.25)

Q3 = df["Monthly Salary (INR)"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - (1.5 * IQR)

upper = Q3 + (1.5 * IQR)    # Calculate the lower and upper bounds for outliers in the "Monthly Salary (INR)" column using the IQR method
```

```python
outliers = df[
    (df["Monthly Salary (INR)"] < lower)
    |
    (df["Monthly Salary (INR)"] > upper)
]

outliers      # Display the rows in the DataFrame that are considered outliers based on the "Monthly Salary (INR)" column using the IQR method
```

```python
df = df[
    (df["Monthly Salary (INR)"] >= lower)
    &
    (df["Monthly Salary (INR)"] <= upper)
]        # Remove outliers from the DataFrame based on the "Monthly Salary (INR)" column using the IQR method
```

# Step 8: Rename and Standardize Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)  # Standardize column names by stripping whitespace, converting to lowercase, and replacing spaces with underscores
```
