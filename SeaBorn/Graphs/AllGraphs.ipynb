## 1. KDE(Kernal Density Estimation) Plot  

What is KDE Plot?   

A KDE Plot (Kernal Density Estimation) is used to show how the data is distributed. 

It Tells us:    

In Which range most of the data is present.

Unlike a Histogram, it does not show the exact count of values. 

Instead, it shows where the data is more concentrated using a smooth curve. 

Example 

Suppose employee salaries are:  

```
20k
25k
28k
30k
31k
32k
35k
80k 
``` 

Histogram says: 

``` 
there are 5 employees in this salary range. 
``` 

KDE says:   

``` 
Most employees have salaries between 30k to 35k.    
``` 

**Probability Denstiy Function (PDF)**  

KDE is based on the Probability Density Function (PDF). 

Instead, it tells us:   
Which range of values is more likely to occur.  

Example     

Instead of saying   

``` 
Salary = 30000  
``` 

PDF says    

``` 
Salary between 29k and 31k has a high chance.   
``` 

This is because salary is continous data, so exact values are less meaningful than ranges.  

**Difference Between Probability and PDF**  

**Probability** 

Shows the chance of one exact value.    

Example 
``` 
Salary = 30000  
``` 

**PDF** 

Shows Which raange has a higher chance  

Example 

``` 
Salary Between 29k and 31k  
``` 

**How KDE Works**

Each data point is covered with a small smooth curve called a Kernel.

Usually, Seaborn uses a Gaussian Kernel (Bell-shaped curve).

Then all these small curves are added together to create one smooth density curve.

This final smooth curve is called the KDE Plot. 

**Why is KDE Smooth?**

Histogram uses bars.

KDE uses a smooth line.

A smooth curve makes it easier to understand the overall pattern of the data.   

**When to Use KDE Plot?**

Use KDE for continuous numerical data such as:

- Salary
- Age
- Customer Spending
- House Price
- Temperature
- Exam Marks

**When Not to Use KDE?**

Do not use KDE for categorical data.

Example

- Male / Female
- Yes / No
- Pass / Fail   

**Advantages**  

- Shows data distribution clearly
- Easy to compare multiple distributions
- Smooth and easy to read
- Shows peaks in the data 

**Limitations** 

- Does not show exact counts
- Can be misleading with very small datasets  

**Company Use** 

- Exploratory Data Analysis (EDA)
- Checking data distribution
- Feature Engineering
- Understanding Normal Distribution
- Machine Learning  

## 2. Box PLot

**What is Box Plot?**   

A Box Plot gives a quick statistical summary of the data.   

It is mainly used to find **outilers.** 

**A Box Plot Shows**    

- Minimum Value
- Maximum Value
- Median
- Quartiles (Q1, Q2, Q3)
- IQR (Interquartile Range)
- Outliers  

**Quartiles**   

Quartiles divide the data into 4 **equal parts.**   

- Q1 → 25%
- Q2 → 50% (Median)
- Q3 → 75%  
- Q4 → 100% 

**Median**  

The middle value of the data.

Median is better than the mean when outliers are present.

**IQR (Interquartile Range)**   

Formula     

``` 
IQR = Q3 - Q1   
``` 

It shows the spread of the middle 50% of the data.  

**Outliers**    


Outliers are values that are much higher or lower than the rest of the data.        

Formula     

``` 
Lower Limit = Q1 - 1.5 * IQR    

Upper Limit = Q3 + 1.5 * IOR    
``` 

Values outside these limits are outliers.   

**Company Use** 

- Employee Salary Analysis
- Fraud Detection
- Medical Data
- Sales Analysis
- Machine Learning

**Advantages**

- Finds outliers easily
- Shows data spread
- Shows median
- Easy to compare groups
- Gives a quick summary

**Limitation**

- It does not show the full distribution or density of the data

## 3. Violin Plot

What is Violin Plot?    

A Violin Plot is a combination of:  

- Box PLot  
- KDE Plot  

It shows both summary and distribution in one graph.    


**It Shows**    

- Median
- Quartiles
- Density
- Distribution Shape
- Peaks
- Spread    

**How to Read It**

If the violin is **wide**, more data is present there.

If the violin is **narrow**, less data is present there.

Example

If most salaries are around 30k, the violin will be wider around 30k.

Difference

**Box Plot**

Shows only the summary.

**Violin Plot**

Shows the summary **plus** the distribution.    

**Company Use**
- Healthcare
- Finance
- Customer Analytics
- Research
- Machine Learning    

**Advantages**
- More informative than a Box Plot
- Shows both summary and density  

**Limitation**

- Can be difficult for beginners to understand. 

## 4. Pair Plot 

What is Pair Plot?  

A Pair Plot automatically shows the realionship between all numercial columns in a dataset. 

It is one of the most useful plots for Exploratory Data Analysis(EDA).  

**Example** 

suppose a dataset has   
- Age   
- Salary    
- Experience    
- marks 

Instead of Creating many separate graphs,   

pair plot creates them **automatically.**   

It compares

- Age vs Salary
- Age vs Experience
- Salary vs Experience
- Salary vs Marks
- Experience vs Marks

All in one figure.  

**Pair Plot Structure**

**Diagonal**

Shows the **distribution** of each feature.

Usually uses

- Histogram
- KDE Plot  

**Off-Diagonal**    

Shows the **realionship** between two features. 

Usually uses    
- Scatter Plot  

**Company Use**
- Exploratory Data Analysis (EDA)
- Feature Selection
- Correlation Analysis
- Finding Relationships
- Pattern Detection

**Advantages**
- One-click analysis
- Saves time
- Shows relationships between all numerical features

**Limitations**
- Slow for large datasets
- Difficult to read when there are many columns 

| Plot            | Main Purpose                                       | Best Use                                          |
| --------------- | -------------------------------------------------- | ------------------------------------------------- |
| **KDE Plot**    | Shows the data distribution and density            | Understanding the distribution of continuous data |
| **Box Plot**    | Shows statistical summary and detects outliers     | Finding outliers and understanding data spread    |
| **Violin Plot** | Shows both distribution and statistical summary    | Comparing the distribution of different groups    |
| **Pair Plot**   | Shows relationships between all numerical features | Quick Exploratory Data Analysis (EDA)             |

**KDE Plot** → Shows where the data is most concentrated.
**Box Plot** → Shows data summary and detects outliers.
**Violin Plot** → Combines Box Plot and KDE Plot in one graph.
**Pair Plot** → Shows relationships between all numerical features in a dataset.

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

Create Dataset

````python
df = pd.DataFrame({
    "Spending": [450, 500, 550, 600, 650, 700, 750, 800, 850, 2500,
                 400, 480, 520, 580, 620, 680, 730, 790, 820, 2100],
    "Gender": ["Male", "Male", "Male", "Male", "Male",
               "Male", "Male", "Male", "Male", "Male",
               "Female", "Female", "Female", "Female", "Female",
               "Female", "Female", "Female", "Female", "Female"]
})
````

Create KDE Plot

````python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="Spending",
    hue="Gender",          # Compare categories
    fill=True,             # Fill the area
    palette="Set2",        # Professional color palette
    alpha=0.35,            # Transparency
    linewidth=2,           # Line thickness
    bw_adjust=1.1,         # Smoothness
    common_norm=False,     # Normalize each category separately
    cut=0                  # Don't extend beyond data range
)

plt.title("Customer Spending Distribution by Gender",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Customer Spending (₹)", fontsize=12)
plt.ylabel("Density", fontsize=12)

plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````

| **Parameter**       | **Purpose**                                                                                                |
| ------------------- | ---------------------------------------------------------------------------------------------------------- |
| `data=df`           | Specifies the dataset to use for the plot.                                                                 |
| `x="Spending"`      | Selects the numerical column to visualize.                                                                 |
| `hue="Gender"`      | Splits the data into different categories using different colors.                                          |
| `fill=True`         | Fills the area under the KDE curve with color.                                                             |
| `palette="Set2"`    | Chooses the color palette for different categories.                                                        |
| `alpha=0.35`        | Controls the transparency of the filled area. Lower value = more transparent.                              |
| `linewidth=2`       | Sets the thickness of the KDE curve.                                                                       |
| `bw_adjust=1.1`     | Controls how smooth the KDE curve looks. Higher value = smoother curve, Lower value = more detailed curve. |
| `common_norm=False` | Calculates the density for each category separately instead of combining them.                             |
| `cut=0`             | Stops the KDE curve from extending beyond the minimum and maximum data values.                             |


````python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Gender",
    y="Spending",
    hue="Gender",
    palette="Set2",
    width=0.6,
    linewidth=2,
    saturation=0.8,
    showmeans=True,
    meanprops={
        "marker": "D",
        "markerfacecolor": "red",
        "markeredgecolor": "black",
        "markersize": 8
    },
    flierprops={
        "marker": "o",
        "markerfacecolor": "red",
        "markeredgecolor": "black",
        "markersize": 7,
        "alpha": 0.8
    }
)

plt.title("Customer Spending Distribution by Gender", fontsize=16, fontweight="bold")
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Customer Spending (₹)", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````

| Parameter          | Purpose                                            |
| ------------------ | -------------------------------------------------- |
| `data=df`          | Dataset used for plotting.                         |
| `x="Gender"`       | Categories shown on the X-axis.                    |
| `y="Spending"`     | Numerical values shown on the Y-axis.              |
| `hue="Gender"`     | Colors each category separately.                   |
| `palette="Set2"`   | Uses a professional color palette.                 |
| `width=0.6`        | Controls the width of each box.                    |
| `linewidth=2`      | Sets the thickness of the box borders.             |
| `saturation=0.8`   | Controls the color intensity of the boxes.         |
| `showmeans=True`   | Displays the mean (average) value inside the plot. |
| `meanprops={...}`  | Customizes the appearance of the mean marker.      |
| `flierprops={...}` | Customizes the appearance of outlier points.       |


````python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Gender",
    y="Spending",
    hue="Gender",
    palette="Set2",
    inner="box",
    split=False,
    density_norm="area",
    linewidth=2,
    saturation=0.8,
    cut=0
)

plt.title("Customer Spending Distribution by Gender",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Gender", fontsize=12)
plt.ylabel("Customer Spending (₹)", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````

| --------------------- | --------------------------------------------------------- |

| `x="Gender"`          | Categories displayed on the X-axis.                       |
| `y="Spending"`        | Numerical values displayed on the Y-axis.                 |
| `hue="Gender"`        | Uses different colors for each category.                  |
| `palette="Set2"`      | Applies a professional color palette.                     |
| `inner="box"`         | Displays a mini Box Plot inside the violin.               |
| `split=False`         | Shows separate violins for each category.                 |
| `density_norm="area"` | Makes all violins have the same area for fair comparison. |
| `linewidth=2`         | Sets the border thickness of the violin.                  |
| `saturation=0.8`      | Controls the color intensity.                             |
| `cut=0`               | Prevents the violin from extending beyond the data range. |


````python
# Create Dataset for pairplot
df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35],
    "Salary": [30000, 35000, 40000, 50000, 65000],
    "Experience": [1, 2, 4, 6, 10],
    "Spending": [500, 700, 900, 1500, 2500],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})
````

## 5. Heatmap   

What is Heatmap?    

A Heatmap is a graph that uses colors to represent values.  

Its main purpose is to show the realtionship between numerical featues. 

Instead of reading many numbers, we can understand the data by looking at the colors.   

**Purpose of Heatmap**

A Heatmap helps us answer questions like:   
- Which features are strongly realted?  
- Which features are weakly realted?    
- Which features have no realtionship?  
- Which features should be used for Machine Learning?   

**Example** 

suppose a company has this employee dataset.    

| Age | Salary | Experience |
| --- | ------ | ---------- |
| 22  | 30000  | 1          |
| 25  | 35000  | 2          |
| 30  | 50000  | 5          |
| 35  | 70000  | 10         |

Wihtout a Heatmap, you have to compare all the numbers manually.    

With a Heatmap, colors immediately show which features are strongly realted.    

**Correlation** 

Heatmap is based on **Correlation.**    

Correlation tells us    

How strongly two numberical variables are realted.  

It only measures the **relationship**, not the cause.   

**Correlation Range**   

Correlation Values always lie between   

```
-1  to  +1  
``` 

**Positive Correlation (+1)**

When one variable increases,

the other variable also increases.

Example

```
Experience ↑

Salary ↑
```

More experience usually means a higher salary.  

**Negative Correlation (-1)**

When one variable increases,

the other variable decreases.

Example

```
Product Price ↑

Sales ↓
```

Higher price often leads to fewer sales.    

**No Correlation (0)**

Both variables have no relationship.

Example

```
Employee ID

Salary
```

Employee ID does not affect salary. 

**Correlation Strength**    

| Correlation      | Meaning                        |
| ---------------- | ------------------------------ |
| **+1.0**         | Perfect Positive Relationship  |
| **+0.7 to +0.9** | Strong Positive Relationship   |
| **+0.4 to +0.6** | Moderate Positive Relationship |
| **+0.1 to +0.3** | Weak Positive Relationship     |
| **0**            | No Relationship                |
| **-0.1 to -0.3** | Weak Negative Relationship     |
| **-0.4 to -0.6** | Moderate Negative Relationship |
| **-0.7 to -0.9** | Strong Negative Relationship   |
| **-1.0**         | Perfect Negative Relationship  |

**How Heatmap Works**

First,

Pandas calculates the Correlation Matrix.

Example 

|            | Age  | Salary | Experience |
| ---------- | ---- | ------ | ---------- |
| Age        | 1.00 | 0.82   | 0.88       |
| Salary     | 0.82 | 1.00   | 0.95       |
| Experience | 0.88 | 0.95   | 1.00       |

Then,   

Heatmap converts these numbers into colors. 

- Strong correlation → Dark color
- Weak correlation → Light color

This makes the data easier to understand.   

**Why Use Heatmap?**

Without Heatmap

You read many numbers.

With Heatmap

You understand the relationships in a few seconds.

This saves time during analysis.    

**When to Use Heatmap?**

Use Heatmap for

- Correlation Analysis
- Feature Selection
- Exploratory Data Analysis (EDA)
- Machine Learning
- Business Analytics
- Finding relationships between numerical columns   

**When Not to Use Heatmap?**

Do not use Heatmap for

- Categorical data
- Text data
- Single numerical column
- Very small datasets with no meaningful relationships  

**Advantages**

- Easy to understand
- Shows relationships quickly
- Uses colors for better visualization
- Helps in feature selection
- Saves analysis time   

**Limitations**

- Works mainly with numerical data
- Large datasets can make the Heatmap crowded
- Correlation does not mean one variable causes the other   

**Company Use**

Companies use Heatmaps for

- Exploratory Data Analysis (EDA)
- Correlation Analysis
- Feature Selection
- Machine Learning
- Customer Analytics
- Sales Analysis
- Financial Analysis
- Business Intelligence 

**Real Example**    

Suppose a bank has these columns.   

```
Age
Income
Loan Amount
Credit Score
``` 

A Heatmap can answer:

- Is Income related to Credit Score?
- Is Loan Amount related to Income?
- Which features have the strongest relationship?
- Which features should be used in the ML model?

Instead of checking every column manually, one Heatmap gives all the relationships at once. 

**Important Note**

A Heatmap does not calculate correlation.

It only visualizes the correlation values.

The correlation is calculated first, and then the Heatmap displays it using colors.

````python
# Create Dataset for heatmap
df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35, 40, 45, 50],
    "Salary": [30000, 35000, 42000, 50000, 65000, 75000, 85000, 95000],
    "Experience": [1, 2, 4, 6, 9, 12, 15, 18],
    "Spending": [500, 700, 900, 1200, 1800, 2200, 2600, 3000],
    "CreditScore": [620, 650, 680, 700, 730, 760, 790, 820]
})
````

````python
plt.figure(figsize=(10, 7))

correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(
    correlation_matrix,
    annot=True,           # Show correlation values
    cmap="coolwarm",      # Professional color palette
    fmt=".2f",            # Show 2 decimal places
    linewidths=0.5,       # Space between cells
    linecolor="white",    # Border color
    square=True,          # Make every cell square
    cbar=True,            # Show color scale
    annot_kws={
        "size": 10,
        "weight": "bold"
    },
    vmin=-1,              # Minimum correlation
    vmax=1                # Maximum correlation
)

plt.title(
    "Feature Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.tight_layout()

plt.show()
````

| Parameter            | Purpose                                               |
| -------------------- | ----------------------------------------------------- |
| `correlation_matrix` | Uses the correlation matrix as input.                 |
| `annot=True`         | Displays the correlation value inside each cell.      |
| `cmap="coolwarm"`    | Applies a professional color theme.                   |
| `fmt=".2f"`          | Shows values with two decimal places.                 |
| `linewidths=0.5`     | Adds space between cells.                             |
| `linecolor="white"`  | Sets the border color of the cells.                   |
| `square=True`        | Makes all cells square-shaped.                        |
| `cbar=True`          | Displays the color bar that explains the color scale. |
| `annot_kws`          | Customizes the annotation text (size and bold style). |
| `vmin=-1`            | Sets the minimum correlation value.                   |
| `vmax=1`             | Sets the maximum correlation value.                   |
